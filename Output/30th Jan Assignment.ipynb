Q1. Write a program to accept percentage from the user and display the grade according to the following criteria: 
Marks	       Grade
>90 	        A
>80 and <= 90	B
>=60 and <= 80	C
below 60	    D


```python
marks=int(input("enter your marks "))
print('marks:',marks)
if marks >90 :
    print("grade A")
elif marks >80 and marks <= 90 :
    print("grade B")
elif marks >= 60 and marks<= 80 :
    print("grade C")
else :
    print("grade C")
```

Q4. Check how many times a given number 
can be divided by 3 before it is less than or equal to 10. 

```python
num = int(input("enter any number"))
print('enter number',num )
counter = 0
while num % 3 == 0 :
    num = num / 3
    counter = counter + 1

print('Times the Number can be divided by Three :',counter)
```

Q6. Use nested while loop to print 3 different pattern.

```python
#PATTERN 1 

a=int(input("enter number of rows "))
while a>0 :
    a-=1
    print("*"*a)
```

```python
#PATTERN 2

a=int(input('enter number of rows '))
b=1 
while a>0 :
    a-=1
    print(" "*a , end ="*"*b +"\n")
    b+=1
    
```

```python
# PATTERN 3 

a=int(input("enter number of rows "))
b=1
while a >0 :
    a-=1
    print(" "*a + "*"*b , end =" "*a +"\n")
    b+=2
```
